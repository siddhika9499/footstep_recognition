```python
# -*- coding: utf-8 -*-
"""footstep_recognition_crnn.ipynb

The "Invisible" Biometric — Footstep Recognition
Identifies WHO is walking based on the sound/vibration of their footsteps.
Architecture : Mel Spectrogram → CNN (feature extraction) → LSTM (temporal rhythm) → Softmax
Replaces     : MobileNetV2 (bird classifier) → Custom CRNN (footstep biometric)
New Concepts : Time-series feature extraction, Recurrent layers, Biometric identification
"""

# ╔══════════════════════════════════════════════════════════════════╗
# ║   FULL PIPELINE — Single Cell (GPU / CPU)                       ║
# ║   Installs → Downloads Dataset → Trains → Saves → Downloads     ║
# ╚══════════════════════════════════════════════════════════════════╝

# ── 1. Install ────────────────────────────────────────────────────
import subprocess
subprocess.run(['pip', 'install', 'kaggle', 'librosa', 'tensorflow',
                'scikit-learn', 'tqdm', 'matplotlib', '-q'], check=True)

# ── 2. Imports ────────────────────────────────────────────────────
import os, json, zipfile, warnings, io, shutil
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import librosa
import librosa.display
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras import layers, models, optimizers, callbacks, regularizers
from collections import Counter
warnings.filterwarnings('ignore')

gpus = tf.config.list_physical_devices('GPU')
print(f'✅ TensorFlow {tf.__version__}  |  GPU: {gpus if gpus else "None (CPU mode)"}')

# ── 3. Kaggle API Setup ───────────────────────────────────────────
KAGGLE_USERNAME = "siddhikapujari"   # ← PASTE YOUR USERNAME
KAGGLE_KEY      = "KGAT_49e2768719aef085dcdcbf986aba9ec8"    # ← PASTE YOUR API KEY

os.makedirs('/root/.kaggle', exist_ok=True)
with open('/root/.kaggle/kaggle.json', 'w') as f:
    json.dump({"username": KAGGLE_USERNAME, "key": KAGGLE_KEY}, f)
os.chmod('/root/.kaggle/kaggle.json', 0o600)
print('✅ Kaggle credentials set')

# ── 4. Download Dataset ───────────────────────────────────────────
# Dataset: "Human Footstep" audio dataset from Kaggle
# This dataset contains footstep recordings from multiple individuals
# walking on different surfaces — perfect for biometric identification.
#
# Primary dataset  : shrutinarvekar/footstep-sound-dataset
# Backup  dataset  : vikrantkamble/human-footstep-signals
#
# The script tries the primary first, falls back to backup automatically.

DATA_ROOT = '/content/footstep_data'
os.makedirs(DATA_ROOT, exist_ok=True)

PRIMARY_DATASET = 'shrutinarvekar/footstep-sound-dataset'
BACKUP_DATASET  = 'vikrantkamble/human-footstep-signals'

def try_download(dataset_slug, dest):
    """Try downloading a Kaggle dataset. Returns True on success."""
    result = subprocess.run(
        ['kaggle', 'datasets', 'download', '-d', dataset_slug, '-p', dest, '-q'],
        capture_output=True, text=True
    )
    return result.returncode == 0

print('⏳ Attempting dataset download (primary)...')
success = try_download(PRIMARY_DATASET, DATA_ROOT)
if not success:
    print('⚠️  Primary dataset not found. Trying backup...')
    success = try_download(BACKUP_DATASET, DATA_ROOT)

if not success:
    # Graceful fallback: generate synthetic demo data so the pipeline still runs
    print('⚠️  No Kaggle dataset found. Generating SYNTHETIC demo data for pipeline testing...')
    print('    (Replace this section with your real dataset for actual biometric results)')

    SYNTH_DIR = os.path.join(DATA_ROOT, 'synthetic_persons')
    os.makedirs(SYNTH_DIR, exist_ok=True)
    SR = 22050
    DUR = 3
    np.random.seed(42)

    person_params = {
        'Person_A': {'step_rate': 1.8, 'weight': 0.9, 'stride': 0.55},
        'Person_B': {'step_rate': 2.2, 'weight': 0.5, 'stride': 0.42},
        'Person_C': {'step_rate': 1.5, 'weight': 1.2, 'stride': 0.65},
        'Person_D': {'step_rate': 2.0, 'weight': 0.7, 'stride': 0.50},
        'Person_E': {'step_rate': 1.6, 'weight': 1.0, 'stride': 0.60},
    }

    for person, params in person_params.items():
        p_dir = os.path.join(SYNTH_DIR, person)
        os.makedirs(p_dir, exist_ok=True)
        for i in range(30):  # 30 samples per person
            t = np.linspace(0, DUR, SR * DUR)
            # Simulate footstep impact pattern (impulse train + decay)
            signal = np.zeros(SR * DUR)
            step_interval = int(SR / params['step_rate'])
            offset = np.random.randint(0, step_interval)
            for step_idx in range(offset, SR * DUR, step_interval):
                decay_len = int(SR * params['stride'])
                if step_idx + decay_len > SR * DUR:
                    decay_len = SR * DUR - step_idx
                decay = np.exp(-np.linspace(0, 6, decay_len)) * params['weight']
                noise = np.random.randn(decay_len) * 0.05
                signal[step_idx:step_idx + decay_len] += (decay + noise) * (0.85 + 0.3 * np.random.rand())
            signal += np.random.randn(SR * DUR) * 0.01  # ambient noise
            signal = signal / (np.max(np.abs(signal)) + 1e-8)  # normalize

            import soundfile as sf
            # If soundfile not available, save as raw numpy
            try:
                import soundfile as sf
                sf.write(os.path.join(p_dir, f'step_{i:03d}.wav'), signal.astype(np.float32), SR)
            except Exception:
                np.save(os.path.join(p_dir, f'step_{i:03d}.npy'), signal.astype(np.float32))

    DATA_ROOT = SYNTH_DIR
    print(f'✅ Synthetic data created: {len(person_params)} persons × 30 samples')
else:
    # Unzip all downloaded archives
    for fname in os.listdir(DATA_ROOT):
        if fname.endswith('.zip'):
            zip_path = os.path.join(DATA_ROOT, fname)
            print(f'📦 Extracting {fname}...')
            with zipfile.ZipFile(zip_path, 'r') as z:
                z.extractall(DATA_ROOT)
    print('✅ Dataset downloaded and extracted')

# ── 5. Discover Person Folders ────────────────────────────────────
# Walk the extracted directory to find the deepest folder structure
# where each subfolder = one person (similar to bird species folders).

def find_person_dir(root):
    """Find the directory whose subfolders each contain audio files."""
    audio_exts = ('.wav', '.mp3', '.ogg', '.flac', '.npy')
    for dirpath, dirnames, filenames in os.walk(root):
        if not dirnames:
            continue
        subfolder_has_audio = []
        for d in dirnames:
            full = os.path.join(dirpath, d)
            has_audio = any(f.lower().endswith(audio_exts)
                            for f in os.listdir(full)
                            if os.path.isfile(os.path.join(full, f)))
            subfolder_has_audio.append(has_audio)
        if sum(subfolder_has_audio) >= 2:
            return dirpath
    return root

PERSON_DIR = find_person_dir(DATA_ROOT)
audio_exts = ('.wav', '.mp3', '.ogg', '.flac', '.npy')

all_persons = sorted([
    d for d in os.listdir(PERSON_DIR)
    if os.path.isdir(os.path.join(PERSON_DIR, d)) and
    any(f.lower().endswith(audio_exts)
        for f in os.listdir(os.path.join(PERSON_DIR, d)))
])

print(f'✅ Found {len(all_persons)} persons in: {PERSON_DIR}')
for p in all_persons:
    n = len([f for f in os.listdir(os.path.join(PERSON_DIR, p))
             if f.lower().endswith(audio_exts)])
    print(f'   {p:25s} — {n} recordings')

# ── 6. Config ─────────────────────────────────────────────────────
SAMPLE_RATE  = 22050   # Hz
DURATION     = 3       # seconds per clip (footsteps are short)
N_MELS       = 64      # mel bins (lower than birds — footsteps are low-freq)
HOP_LENGTH   = 256     # smaller hop = more temporal frames for LSTM
N_FFT        = 1024
TIME_STEPS   = int(np.ceil(SAMPLE_RATE * DURATION / HOP_LENGTH))  # frames for LSTM
BATCH_SIZE   = 16      # smaller batch — fewer samples than bird dataset
AUTOTUNE     = tf.data.AUTOTUNE

print(f'\n📐 CRNN Input shape: ({TIME_STEPS} time steps, {N_MELS} mel bins)')
print(f'   Batch size : {BATCH_SIZE}')
print(f'   Duration   : {DURATION}s @ {SAMPLE_RATE}Hz')

# ── 7. Audio → Mel Spectrogram (CRNN format) ──────────────────────
#
# KEY DIFFERENCE from bird classifier:
#   Bird classifier  → (IMG_SIZE, IMG_SIZE, 3)  — image format for CNN
#   Footstep CRNN    → (TIME_STEPS, N_MELS)     — sequence format for LSTM
#
# We keep TIME as the first axis so LSTM can read frames left-to-right
# and learn the RHYTHM of each person's walk (pace, cadence, weight).

def audio_to_crnn_input(file_path):
    """
    Convert audio file → (TIME_STEPS, N_MELS) mel spectrogram.
    Returns None on error.
    """
    try:
        # Handle .npy files (synthetic data)
        if file_path.endswith('.npy'):
            y = np.load(file_path).astype(np.float32)
            if len(y) < SAMPLE_RATE * DURATION:
                y = np.pad(y, (0, SAMPLE_RATE * DURATION - len(y)))
            else:
                y = y[:SAMPLE_RATE * DURATION]
        else:
            y, _ = librosa.load(file_path, sr=SAMPLE_RATE,
                                 duration=DURATION, mono=True)

        # Pad / trim to exact length
        target_len = SAMPLE_RATE * DURATION
        if len(y) < target_len:
            y = np.pad(y, (0, target_len - len(y)))
        else:
            y = y[:target_len]

        # Mel spectrogram
        mel = librosa.feature.melspectrogram(
            y=y, sr=SAMPLE_RATE,
            n_mels=N_MELS,
            hop_length=HOP_LENGTH,
            n_fft=N_FFT,
            fmin=20,    # footsteps: 20 Hz - 4 kHz (low frequency focus)
            fmax=4000
        )
        mel_db = librosa.power_to_db(mel, ref=np.max)

        # Normalize to [0, 1]
        mel_norm = (mel_db - mel_db.min()) / (mel_db.max() - mel_db.min() + 1e-8)

        # Shape: (N_MELS, TIME_STEPS) → transpose → (TIME_STEPS, N_MELS)
        # LSTM reads across time, so TIME must be axis 0
        crnn_input = mel_norm.T  # (TIME_STEPS, N_MELS)

        # Ensure consistent shape
        if crnn_input.shape[0] < TIME_STEPS:
            pad_len = TIME_STEPS - crnn_input.shape[0]
            crnn_input = np.pad(crnn_input, ((0, pad_len), (0, 0)))
        else:
            crnn_input = crnn_input[:TIME_STEPS, :]

        return crnn_input.astype(np.float32)

    except Exception as e:
        return None

# Quick sanity test
test_files = [os.path.join(PERSON_DIR, all_persons[0], f)
              for f in os.listdir(os.path.join(PERSON_DIR, all_persons[0]))
              if f.lower().endswith(audio_exts)][:1]
if test_files:
    test_out = audio_to_crnn_input(test_files[0])
    if test_out is not None:
        print(f'✅ Preprocessing check: input shape = {test_out.shape}  (expected: ({TIME_STEPS}, {N_MELS}))')

# ── 8. Build Dataset ──────────────────────────────────────────────
label_to_idx = {p: i for i, p in enumerate(all_persons)}
X, y_labels = [], []

for person in tqdm(all_persons, desc='🔊 Building dataset'):
    person_path = os.path.join(PERSON_DIR, person)
    files = [f for f in os.listdir(person_path) if f.lower().endswith(audio_exts)]

    for fname in files:
        fpath = os.path.join(person_path, fname)
        features = audio_to_crnn_input(fpath)
        if features is not None:
            X.append(features)
            y_labels.append(label_to_idx[person])

X        = np.array(X, dtype=np.float32)   # (N, TIME_STEPS, N_MELS)
y_labels = np.array(y_labels, dtype=np.int32)

print(f'\n✅ Dataset built:')
print(f'   Total samples : {len(X)}')
print(f'   Array shape   : {X.shape}   (samples × time_steps × mel_bins)')
print(f'   Label counts  : {dict(Counter(y_labels))}')

# ── 9. Filter & Remap Labels ──────────────────────────────────────
counts     = np.bincount(y_labels)
valid_mask = counts[y_labels] >= 2       # need ≥2 samples to stratify split
X_f        = X[valid_mask]
y_f        = y_labels[valid_mask]

unique_lbls   = np.unique(y_f)
remap         = {old: new for new, old in enumerate(unique_lbls)}
y_remapped    = np.array([remap[l] for l in y_f])
persons_final = [all_persons[i] for i in unique_lbls]
NUM_CLASSES   = len(persons_final)

print(f'\n✅ After filtering singletons: {NUM_CLASSES} persons, {len(X_f)} samples')

# ── 10. Train / Validation Split ──────────────────────────────────
X_train, X_val, y_train, y_val = train_test_split(
    X_f, y_remapped,
    test_size=0.2,
    random_state=42,
    stratify=y_remapped
)
print(f'   Train : {len(X_train)} samples')
print(f'   Val   : {len(X_val)} samples')

# ── 11. tf.data Pipelines ─────────────────────────────────────────
def augment_footstep(features, label):
    """
    Temporal augmentation — critical for footstep biometrics.
    We augment in the time domain to preserve rhythm patterns.
    """
    # Time masking: randomly zero out 10% of frames (simulates missed step)
    time_mask_len = tf.cast(tf.shape(features)[0], tf.float32) * 0.10
    time_mask_len = tf.cast(time_mask_len, tf.int32)
    mask_start    = tf.random.uniform([], 0, tf.shape(features)[0] - time_mask_len, dtype=tf.int32)
    mask = tf.concat([
        tf.ones([mask_start, N_MELS]),
        tf.zeros([time_mask_len, N_MELS]),
        tf.ones([tf.shape(features)[0] - mask_start - time_mask_len, N_MELS])
    ], axis=0)
    features = features * mask

    # Frequency masking: randomly zero 10% of mel bins
    freq_mask_len = tf.cast(N_MELS * 0.10, tf.int32)
    freq_start    = tf.random.uniform([], 0, N_MELS - freq_mask_len, dtype=tf.int32)
    freq_mask = tf.concat([
        tf.ones([TIME_STEPS, freq_start]),
        tf.zeros([TIME_STEPS, freq_mask_len]),
        tf.ones([TIME_STEPS, N_MELS - freq_start - freq_mask_len])
    ], axis=1)
    features = features * freq_mask

    # Small noise injection (simulates different floor surfaces)
    features = features + tf.random.normal(tf.shape(features), stddev=0.01)
    features = tf.clip_by_value(features, 0.0, 1.0)

    return features, label

train_ds = (tf.data.Dataset.from_tensor_slices((X_train, y_train))
            .shuffle(500)
            .map(augment_footstep, num_parallel_calls=AUTOTUNE)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

val_ds   = (tf.data.Dataset.from_tensor_slices((X_val, y_val))
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

print('✅ tf.data pipelines ready (with SpecAugment-style augmentation)')

# ── 12. Build CRNN Model ──────────────────────────────────────────
#
# Architecture explanation:
#
#   Input  : (TIME_STEPS, N_MELS)
#              — each timestep = one mel frame, read left to right
#
#   Conv1D blocks : Extract LOCAL patterns (single footstep impact shape,
#                   heel-strike signature, surface texture)
#
#   Bidirectional LSTM : Learn TEMPORAL patterns across the whole sequence
#                        (cadence, stride rhythm, weight distribution over time)
#                        Bidirectional = reads forward AND backward
#                        so it captures context around each step
#
#   Dense head : Classification into N persons
#
# Why NOT MobileNetV2 here?
#   MobileNetV2 treats the spectrogram as a static image — it loses all
#   temporal ordering. Walking rhythm is fundamentally sequential, so we
#   need a recurrent layer (LSTM) that respects the time axis.

def build_crnn(num_classes, time_steps, n_mels):
    inp = layers.Input(shape=(time_steps, n_mels), name='mel_input')

    # ── CNN Block 1: Local feature extraction ──
    x = layers.Conv1D(64, kernel_size=5, padding='same',
                      kernel_regularizer=regularizers.l2(1e-4),
                      name='conv1')(inp)
    x = layers.BatchNormalization(name='bn1')(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling1D(pool_size=2, name='pool1')(x)
    x = layers.Dropout(0.2)(x)

    # ── CNN Block 2: Mid-level patterns ──
    x = layers.Conv1D(128, kernel_size=3, padding='same',
                      kernel_regularizer=regularizers.l2(1e-4),
                      name='conv2')(x)
    x = layers.BatchNormalization(name='bn2')(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling1D(pool_size=2, name='pool2')(x)
    x = layers.Dropout(0.2)(x)

    # ── CNN Block 3: High-level temporal patterns ──
    x = layers.Conv1D(128, kernel_size=3, padding='same',
                      kernel_regularizer=regularizers.l2(1e-4),
                      name='conv3')(x)
    x = layers.BatchNormalization(name='bn3')(x)
    x = layers.Activation('relu')(x)
    x = layers.Dropout(0.2)(x)

    # ── Bidirectional LSTM: Temporal rhythm learning ──
    # return_sequences=True on first LSTM → pass full sequence to second LSTM
    x = layers.Bidirectional(
        layers.LSTM(128, return_sequences=True, dropout=0.3,
                    recurrent_dropout=0.1, name='lstm1'),
        name='bilstm1'
    )(x)

    # Second LSTM: compress sequence to single context vector
    x = layers.Bidirectional(
        layers.LSTM(64, return_sequences=False, dropout=0.3,
                    recurrent_dropout=0.1, name='lstm2'),
        name='bilstm2'
    )(x)

    # ── Classification Head ──
    x = layers.Dense(128, activation='relu',
                     kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.3)(x)
    out = layers.Dense(num_classes, activation='softmax', name='predictions')(x)

    model = models.Model(inp, out, name='Footstep_CRNN')
    return model

crnn_model = build_crnn(NUM_CLASSES, TIME_STEPS, N_MELS)
crnn_model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print(f'\n✅ CRNN Model built:')
crnn_model.summary()
print(f'\n   Input  : (batch, {TIME_STEPS}, {N_MELS}) — time × frequency')
print(f'   Output : (batch, {NUM_CLASSES}) — one score per person')
print(f'   Params : {crnn_model.count_params():,}')

# ── 13. Callbacks ─────────────────────────────────────────────────
cb_list = [
    callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=8,
        restore_best_weights=True,
        verbose=1
    ),
    callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=4,
        min_lr=1e-6,
        verbose=1
    ),
    callbacks.ModelCheckpoint(
        'best_footstep_model.h5',
        monitor='val_accuracy',
        save_best_only=True,
        verbose=1
    ),
    callbacks.CSVLogger('training_log.csv')
]

# ── 14. Phase 1 — Train full CRNN ─────────────────────────────────
print('\n── Phase 1: Training full CRNN (all layers) ──')
history1 = crnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=40,
    callbacks=cb_list,
    verbose=1
)
print('✅ Phase 1 complete!')

# ── 15. Phase 2 — Fine-tune with lower LR ─────────────────────────
# Reduce LR and retrain — helps LSTM converge to person-specific rhythms
crnn_model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
print('\n── Phase 2: Fine-tuning at lower learning rate (1e-4) ──')
history2 = crnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=cb_list,
    verbose=1
)
print('✅ Phase 2 complete!')

# ── 16. Training Curves ───────────────────────────────────────────
acc      = history1.history['accuracy']     + history2.history['accuracy']
val_acc  = history1.history['val_accuracy'] + history2.history['val_accuracy']
loss     = history1.history['loss']         + history2.history['loss']
val_loss = history1.history['val_loss']     + history2.history['val_loss']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Footstep CRNN — Training Curves', fontsize=14, fontweight='bold')

for ax, tr, vl, title, ylabel in zip(
    axes,
    [acc, loss],
    [val_acc, val_loss],
    ['Accuracy', 'Loss'],
    ['Accuracy', 'Loss']
):
    ax.plot(tr, label='Train',      color='steelblue', linewidth=2)
    ax.plot(vl, label='Validation', color='coral',     linewidth=2)
    ax.axvline(x=len(history1.history['accuracy']),
               color='gray', linestyle='--', alpha=0.7, label='Phase 2 start')
    ax.set_title(f'Model {title}', fontsize=12)
    ax.set_xlabel('Epoch')
    ax.set_ylabel(ylabel)
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_curves_footstep.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'🏆 Best val accuracy: {max(val_acc)*100:.2f}%')

# ── 17. Mel Spectrogram Visualization ────────────────────────────
# Show how different persons have visually distinct footstep patterns
fig, axes = plt.subplots(1, min(4, NUM_CLASSES), figsize=(16, 4))
if NUM_CLASSES == 1:
    axes = [axes]
axes = axes if hasattr(axes, '__len__') else [axes]

fig.suptitle('Footstep Mel Spectrograms — Per Person', fontsize=13, fontweight='bold')

for i, person in enumerate(persons_final[:4]):
    p_dir   = os.path.join(PERSON_DIR, person)
    p_files = [f for f in os.listdir(p_dir) if f.lower().endswith(audio_exts)]
    if not p_files:
        continue
    sample  = audio_to_crnn_input(os.path.join(p_dir, p_files[0]))
    if sample is None:
        continue
    # sample shape: (TIME_STEPS, N_MELS) → transpose for display
    axes[i].imshow(sample.T, aspect='auto', origin='lower',
                   cmap='magma', interpolation='nearest')
    axes[i].set_title(person, fontsize=11)
    axes[i].set_xlabel('Time frames →')
    axes[i].set_ylabel('Mel bins')

plt.tight_layout()
plt.savefig('footstep_spectrograms.png', dpi=150, bbox_inches='tight')
plt.show()

# ── 18. Save Model & Labels ───────────────────────────────────────
best_model = tf.keras.models.load_model('best_footstep_model.h5')
best_model.save('footstep_crnn_final.h5')

model_config = {
    "persons"    : persons_final,
    "num_classes": NUM_CLASSES,
    "sample_rate": SAMPLE_RATE,
    "duration"   : DURATION,
    "n_mels"     : N_MELS,
    "hop_length" : HOP_LENGTH,
    "n_fft"      : N_FFT,
    "time_steps" : TIME_STEPS,
    "architecture": "CRNN (Conv1D × 3 + BiLSTM × 2)"
}

with open('footstep_model_config.json', 'w') as f:
    json.dump(model_config, f, indent=2)

print(f'\n✅ Model saved   : {os.path.getsize("footstep_crnn_final.h5")/1e6:.1f} MB')
print(f'✅ Config saved  : {NUM_CLASSES} persons — {persons_final}')

# ── 19. Download Files ────────────────────────────────────────────
from google.colab import files
files.download('footstep_crnn_final.h5')
files.download('footstep_model_config.json')
files.download('training_curves_footstep.png')

print()
print('=' * 55)
print('  DONE — check your browser downloads!')
print('=' * 55)
print(f'  Persons   : {NUM_CLASSES}')
print(f'  Val Acc   : {max(val_acc)*100:.2f}%')
print(f'  Params    : {crnn_model.count_params():,}')
print(f'  Architecture : Conv1D × 3 → BiLSTM × 2 → Dense')
print('=' * 55)


# ╔══════════════════════════════════════════════════════════════════╗
# ║   INFERENCE UI — Run this in a separate cell after training     ║
# ╚══════════════════════════════════════════════════════════════════╝

import tensorflow as tf
import numpy as np
import librosa
import json
import io
import IPython.display as ipd
from ipywidgets import VBox, HBox, Button, Output, HTML, FileUpload, Layout
import ipywidgets as widgets

# ── Load model & config ──────────────────────────────────────────
model = tf.keras.models.load_model('footstep_crnn_final.h5')

with open('footstep_model_config.json', 'r') as f:
    cfg = json.load(f)

persons_list = cfg['persons']
SAMPLE_RATE  = cfg['sample_rate']
DURATION     = cfg['duration']
N_MELS       = cfg['n_mels']
HOP_LENGTH   = cfg['hop_length']
N_FFT        = cfg['n_fft']
TIME_STEPS   = cfg['time_steps']

# ── Preprocessing (inference version) ────────────────────────────
def preprocess_audio_bytes(audio_bytes):
    """Audio bytes → CRNN input tensor."""
    y, _ = librosa.load(io.BytesIO(audio_bytes), sr=SAMPLE_RATE,
                        duration=DURATION, mono=True)
    target_len = SAMPLE_RATE * DURATION
    y = np.pad(y, (0, max(0, target_len - len(y))))[:target_len]

    mel = librosa.feature.melspectrogram(
        y=y, sr=SAMPLE_RATE,
        n_mels=N_MELS, hop_length=HOP_LENGTH, n_fft=N_FFT,
        fmin=20, fmax=4000
    )
    mel_db = librosa.power_to_db(mel, ref=np.max)
    mel_norm = (mel_db - mel_db.min()) / (mel_db.max() - mel_db.min() + 1e-8)

    crnn_input = mel_norm.T  # (TIME_STEPS, N_MELS)
    if crnn_input.shape[0] < TIME_STEPS:
        crnn_input = np.pad(crnn_input, ((0, TIME_STEPS - crnn_input.shape[0]), (0, 0)))
    else:
        crnn_input = crnn_input[:TIME_STEPS, :]

    return crnn_input.astype(np.float32)

def predict_person(audio_bytes):
    """Returns top-3 person predictions with confidence scores."""
    features = preprocess_audio_bytes(audio_bytes)
    features = np.expand_dims(features, axis=0)  # (1, TIME_STEPS, N_MELS)
    preds    = model.predict(features, verbose=0)[0]
    top3     = np.argsort(preds)[-3:][::-1]
    return [(persons_list[i], float(preds[i])) for i in top3]

# ── UI Styles ─────────────────────────────────────────────────────
CARD_STYLE = """
<style>
  .foot-card {
    font-family: 'Segoe UI', sans-serif;
    background: #f8fafb;
    border: 1px solid #e2e8f0;
    border-radius: 12px;
    padding: 20px 24px;
    max-width: 520px;
  }
  .foot-card h2 { margin: 0 0 4px; font-size: 18px; color: #1a202c; }
  .foot-card .sub { font-size: 13px; color: #718096; margin-bottom: 16px; }
  .foot-result-top {
    background: #ebf0f8;
    border-left: 4px solid #3182ce;
    border-radius: 8px;
    padding: 14px 16px;
    margin-bottom: 12px;
  }
  .foot-result-top .name { font-size: 20px; font-weight: 700; color: #1a365d; }
  .foot-result-top .conf { font-size: 14px; color: #2b4c7e; margin-top: 2px; }
  .foot-result-top .tag  { font-size: 11px; background: #bee3f8; color: #2c5282;
                           padding: 2px 8px; border-radius: 20px; display:inline-block;
                           margin-top: 6px; }
  .bar-row   { display: flex; align-items: center; gap: 10px; margin-bottom: 8px; }
  .bar-label { width: 150px; font-size: 13px; color: #2d3748;
               white-space: nowrap; overflow: hidden; text-overflow: ellipsis; }
  .bar-track { flex: 1; background: #e2e8f0; border-radius: 6px; height: 10px; overflow: hidden; }
  .bar-fill  { height: 100%; border-radius: 6px; background: #3182ce; }
  .bar-pct   { font-size: 12px; color: #4a5568; width: 42px; text-align: right; }
  .divider   { border: none; border-top: 1px solid #e2e8f0; margin: 14px 0; }
  .section-label { font-size: 11px; font-weight: 600; color: #a0aec0;
                   text-transform: uppercase; letter-spacing: 0.06em; margin-bottom: 10px; }
</style>
"""

def result_html(predictions):
    top_name, top_conf = predictions[0]
    pct_top = round(top_conf * 100, 1)

    confidence_label = ("High confidence" if pct_top > 70
                        else "Moderate confidence" if pct_top > 45
                        else "Low confidence — try a longer clip")

    bars = ""
    for name, conf in predictions:
        pct = round(conf * 100, 1)
        bars += f"""
        <div class="bar-row">
          <span class="bar-label" title="{name}">{name}</span>
          <div class="bar-track">
            <div class="bar-fill" style="width:{pct}%"></div>
          </div>
          <span class="bar-pct">{pct}%</span>
        </div>"""

    return f"""
    {CARD_STYLE}
    <div class="foot-card">
      <h2>Footstep Identified</h2>
      <p class="sub">Based on gait rhythm & acoustic biometric analysis</p>
      <div class="foot-result-top">
        <div class="name">{top_name}</div>
        <div class="conf">Confidence: {pct_top}%</div>
        <div class="tag">{confidence_label}</div>
      </div>
      <hr class="divider">
      <div class="section-label">Top 3 predictions</div>
      {bars}
    </div>"""

# ── Build Widget UI ───────────────────────────────────────────────
title_html = HTML(f"""
<div style="font-family:'Segoe UI',sans-serif; max-width:520px; margin-bottom:8px;">
  <h2 style="margin:0 0 4px; font-size:22px; color:#1a202c;">Footstep Biometric Identifier</h2>
  <p style="margin:0; font-size:14px; color:#718096;">
    Upload a <code>.wav</code> or <code>.mp3</code> footstep recording to identify the person.<br>
    <b>Persons in model:</b> {', '.join(persons_list)}
  </p>
</div>
""")

uploader  = FileUpload(accept='.wav,.mp3,.ogg', multiple=False,
                       description='Choose audio',
                       layout=Layout(width='200px'))
run_btn   = Button(description='Identify Person', button_style='primary',
                   icon='user', layout=Layout(width='165px', height='36px'))
clear_btn = Button(description='Clear', button_style='',
                   icon='times', layout=Layout(width='90px', height='36px'))
status    = HTML("")
out       = Output()

def on_identify(b):
    out.clear_output()
    status.value = ""
    if not uploader.value:
        status.value = '<span style="color:#c53030;font-size:13px;">No file uploaded.</span>'
        return
    file_info  = list(uploader.value.values())[0]
    audio_data = file_info['content']
    fname      = file_info['metadata']['name']
    status.value = f'<span style="color:#3182ce;font-size:13px;">Analysing <b>{fname}</b>…</span>'
    run_btn.disabled = True
    try:
        predictions = predict_person(bytes(audio_data))
        with out:
            ipd.display(HTML(result_html(predictions)))
        status.value = '<span style="color:#276749;font-size:13px;">Done.</span>'
    except Exception as e:
        status.value = f'<span style="color:#c53030;font-size:13px;">Error: {e}</span>'
    finally:
        run_btn.disabled = False

def on_clear(b):
    out.clear_output()
    status.value = ""

run_btn.on_click(on_identify)
clear_btn.on_click(on_clear)

ui = VBox([
    title_html,
    HBox([uploader, run_btn, clear_btn],
         layout=Layout(gap='10px', align_items='center')),
    status,
    out
], layout=Layout(gap='10px', padding='16px 0'))

ipd.display(ui)
```
